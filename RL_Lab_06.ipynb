```python

import gymnasium as gym
import random

# Create environment
env = gym.make("CliffWalking-v1")

# Number of episodes
episodes = 10

episode_rewards = []

for episode in range(1, episodes + 1):

    # Reset environment
    state, info = env.reset()

    total_reward = 0
    terminated = False
    truncated = False

    while not (terminated or truncated):

        # Get available actions
        actions = list(range(env.action_space.n))

        # Select random action
        action = random.choice(actions)

        # Execute action
        next_state, reward, terminated, truncated, info = env.step(action)

        # Add reward
        total_reward += reward

        # Move to next state
        state = next_state

    # Store episode reward
    episode_rewards.append(total_reward)

    print(f"Episode {episode}: Total Reward = {total_reward}")

# Calculate average reward
average_reward = sum(episode_rewards) / len(episode_rewards)

print("\nAverage Reward =", average_reward)

env.close()
```

```python
import gymnasium as gym
import random


def random_agent(environment_name, episodes=10):

    # Create environment
    env = gym.make(environment_name)

    episode_rewards = []

    print("\n===================================")
    print("Environment:", environment_name)
    print("===================================")

    for episode in range(1, episodes + 1):

        # Reset environment
        state, info = env.reset()

        total_reward = 0
        terminated = False
        truncated = False

        while not (terminated or truncated):

            # Identify available actions
            actions = list(range(env.action_space.n))

            # Select random action
            action = random.choice(actions)

            # Execute action
            next_state, reward, terminated, truncated, info = env.step(action)

            # Add reward
            total_reward += reward

            # Update state
            state = next_state
            episode_rewards.append(total_reward)

        print(
            f"Episode {episode}: "
            f"Total Reward = {total_reward}"
        )

    # Average reward
    average_reward = sum(episode_rewards) / len(episode_rewards)

    print("\nAverage Reward:", average_reward)

    env.close()


# Test CliffWalking
random_agent("CliffWalking-v1", episodes=10)

# Test Taxi
random_agent("Taxi-v4", episodes=10)
```
